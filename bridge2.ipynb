```python
import json
import pandas as pd

file_path = 'bridge_results_v2.json'

with open(file_path, 'r') as f:
    results = json.load(f)

# Display the top-level keys and their types
print("Top-level structure:")
for key, value in results.items():
    if isinstance(value, list):
        print(f"- '{key}': list of {len(value)} items")
    elif isinstance(value, dict):
        print(f"- '{key}': dict with keys {list(value.keys())}")
    else:
        print(f"- '{key}': {type(value).__name__} = {value}")

print("\n" + "="*50 + "\n")

# Attempt to display lists of dictionaries as DataFrames for better readability
for key, value in results.items():
    if isinstance(value, list) and len(value) > 0 and isinstance(value[0], dict):
        print(f"\nDataframe preview for '{key}':")
        display(pd.DataFrame(value).head())
```

```python
import pandas as pd
from collections import Counter

triple_tokens = []

# Iterate through each prompt to extract tokens where the zone is 'triple'
for item in results['per_prompt']:
    toks = item['toks']
    zones = item['zone']
    for token, zone in zip(toks, zones):
        if zone == 'triple':
            triple_tokens.append(token)

# Count token frequencies to identify patterns
token_counts = Counter(triple_tokens)

print(f"Total tokens in the 'triple' zone: {len(triple_tokens)}")
print("\nTop 20 most common 'triple' tokens:")

# Display as a dataframe for a clean table view
df_triple_tokens = pd.DataFrame(token_counts.most_common(20), columns=['Token', 'Frequency'])
display(df_triple_tokens)
```

```python
nero_prompts = []

# Search for 'Nero' (with or without a leading space) in the token lists
for item in results['per_prompt']:
    if 'Nero' in item['toks'] or ' Nero' in item['toks']:
        # Join the tokens to reconstruct the text
        text = "".join(item['toks'])
        nero_prompts.append(text)

print(f"Found {len(nero_prompts)} prompts containing 'Nero'.")
print("\nHere are the first 3 examples to show the context:\n")

for i, prompt in enumerate(nero_prompts[:3]):
    print(f"Example {i+1}:")
    print(prompt)
    print("-" * 80)
```

```python
import pandas as pd
from collections import Counter

all_tokens = []
triple_tokens = []

# Gather all tokens and triple zone tokens
for item in results['per_prompt']:
    toks = item['toks']
    zones = item['zone']
    all_tokens.extend(toks)
    for t, z in zip(toks, zones):
        if z == 'triple':
            triple_tokens.append(t)

all_counts = Counter(all_tokens)
triple_counts = Counter(triple_tokens)

total_all = sum(all_counts.values())
total_triple = sum(triple_counts.values())

data = []
for token, t_count in triple_counts.items():
    # Filter out very rare tokens to avoid noisy lift calculations
    if t_count > 15:
        a_count = all_counts[token]
        t_freq = t_count / total_triple
        a_freq = a_count / total_all
        lift = t_freq / a_freq

        data.append({
            'Token': repr(token),  # use repr to clearly see spaces/newlines
            'Triple_Count': t_count,
            'All_Count': a_count,
            'Lift (Multiplier)': lift
        })

df_lift = pd.DataFrame(data)

print("Top 20 tokens with the HIGHEST LIFT (most disproportionately over-represented in 'triple'):")
display(df_lift.sort_values('Lift (Multiplier)', ascending=False).head(20))

print("\nWait, what about the most frequent words? Let's look at their Lift:")
display(df_lift.sort_values('Triple_Count', ascending=False).head(10))
```
